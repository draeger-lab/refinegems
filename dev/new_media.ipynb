```python
import pathlib
import pandas as pd

from refinegems.utility.databases import initialise_database
from refinegems.classes.medium import Medium, enter_medium_into_db, updated_db_to_schema
```

```python
# Delete data.db & initialise newly to get the latest version
initialise_database()
```

```python
# Set the directory containing the new media TSV files
dirname = './new_media/'
current_dir = pathlib.Path(dirname)

# Iterate over all TSV files, generate Medium object and insert into database
for file in current_dir.rglob("*.tsv"):
    medium_name = file.stem.upper()
    substance_table = pd.read_csv(file, sep="\t")
    medium_description = substance_table['description'].iloc[0]
    medium_doi = substance_table['doi'].iloc[0]
    substance_table = substance_table.drop(columns=['description', 'doi'])
    medium = Medium(
        name=medium_name,
        substance_table=substance_table,
        description=medium_description,
        doi=medium_doi
    )
    enter_medium_into_db(medium)
```

```python
# Manual/automatic update of substance2db for new substances before the next step required!

# Update database schema after all new media have been entered
updated_db_to_schema('../src/refinegems/data/database')
```
